```python
import numpy as np
data = {
"EmpID": [101, 102, 103, 104, 105],
"Name": ["Amit", "Bhavna", None, "Divya", "Eshan"],
"Age": [28, np.nan, 29, None, 30],
"Salary": [45000, 55000, None, 50000, np.nan]
}
df = pd.DataFrame(data)
print(df)
print(df.isnull()) # True where value is missing
print(df.isnull().sum()) # count of missing values p
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x, y)
plt.xlabel("Day")
plt.ylabel("Sales (units)")
plt.title("Sales over Time")
plt.show()
```

```python
plt.plot(x, y)
plt.grid(True) # background grid
plt.xlim(0, 6) # X-axis range
plt.ylim(0, 35) # Y-axis range
plt.show()
```

```python
plt.plot(x, y)
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([0, 10, 20, 30])
plt.show()
```

```python
plt.plot(x, y, label="Sales Trend")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x, y, label="Sales Trend",
color="blue", marker="o", linestyle="--")
plt.xlabel("Time (Days)")
plt.ylabel("Sales (Units)")
plt.title("Line Graph Example — Sales over Time")
plt.grid(True)
plt.xlim(0, 6)
plt.ylim(0, 35)
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([0, 10, 20, 30])
plt.legend()
plt.show()
```
